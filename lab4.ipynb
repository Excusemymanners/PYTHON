# Laborator 4 – Analiza datelor cu Pandas & Matplotlib

Set de date: jucători FIFA (`data.csv`)

```python
import pandas as pd
import matplotlib.pyplot as plt
import matplotlib
matplotlib.use('Agg')

# Ex 1 – Incarcam si afisam intregul set de date
df = pd.read_csv('data.csv')
print(f"Shape: {df.shape}")
df.head()
```

```python
# Ex 2 – Primii 10 jucatori cu varsta > 40
df_over40 = df[df['Age'] > 40].head(10)
print(f"Jucatori cu varsta > 40: {len(df[df['Age'] > 40])}")
print(df_over40[['Name','Age','Overall','Club']].to_string(index=False))
```

```python
# Ex 3 – Jucatori cu Overall >= 85 si Age < 25
df_young_good = df[(df['Overall'] >= 85) & (df['Age'] < 25)]
print(f"Jucatori cu Overall>=85 si Age<25: {len(df_young_good)}")
print(df_young_good[['Name','Age','Overall']].to_string(index=False))
```

```python
# Ex 4 – Sortare dupa Skill Moves descrescator
df_sorted = df.sort_values('Skill Moves', ascending=False)
print(df_sorted[['Name','Skill Moves','Overall']].head(10).to_string(index=False))
```

```python
# Ex 5 – Jucatori cu contract pana in 2021
df_2021 = df[df['Contract Valid Until'] <= 2021]
print(f"Jucatori cu contract pana in 2021: {len(df_2021)}")
print(df_2021[['Name','Club','Contract Valid Until']].head(10).to_string(index=False))
```

```python
# Ex 6 – Randuri, coloane, jucatori unici
print(f"Randuri: {df.shape[0]}, Coloane: {df.shape[1]}")
print(f"Jucatori unici: {df['Name'].nunique()}")
```

```python
# Ex 7 – Top 5 nationalitati
top5 = df['Nationality'].value_counts().head(5)
print("Top 5 nationalitati:")
print(top5.to_string())
```

```python
# Ex 8 – Pie chart top 5 nationalitati
fig, ax = plt.subplots(figsize=(7,7))
top5 = df['Nationality'].value_counts().head(5)
ax.pie(top5.values, labels=top5.index, autopct='%1.1f%%', startangle=140)
ax.set_title('Proportia jucatorilor pe nationalitati (Top 5)')
plt.tight_layout()
plt.savefig('lab4_pie.png', dpi=80)
plt.show()
print("Grafic salvat: lab4_pie.png")
```

```python
# Ex 9 – Media SprintSpeed si Acceleration per nationalitate
df_num = df[['Nationality','SprintSpeed','Acceleration']].copy()
media_nat = df_num.groupby('Nationality')[['SprintSpeed','Acceleration']].mean().round(2)
print(media_nat.sort_values('SprintSpeed', ascending=False).head(10).to_string())
```

```python
# Ex 10 – Completare valori lipsa Position cu 'Unknown'
df['Position'] = df['Position'].replace('', pd.NA).fillna('Unknown')
print(f"Valori lipsa Position dupa completare: {df['Position'].isna().sum()}")
print(df['Position'].value_counts().head())
```

```python
# Ex 11 – Clubul cu cea mai mare medie de Overall
media_club = df.groupby('Club')['Overall'].mean().round(2).sort_values(ascending=False)
print("Clubul cu cea mai mare medie Overall:")
print(media_club.head(5).to_string())
```

```python
# Ex 12 – Cati jucatori au Value > Wage
df['Value'] = pd.to_numeric(df['Value'], errors='coerce')
df['Wage'] = pd.to_numeric(df['Wage'], errors='coerce')
n_val_gt_wage = (df['Value'] > df['Wage']).sum()
print(f"Jucatori cu Value > Wage: {n_val_gt_wage}")
```

```python
# Ex 13 – Coloana 'is_underpaid'
df['is_underpaid'] = df['Wage'] < df['Value'] / 100
print(f"Jucatori sub-platiti: {df['is_underpaid'].sum()}")
print(df[df['is_underpaid']][['Name','Wage','Value']].head(5).to_string(index=False))
```

```python
# Ex 14 – Scor general
df['Score'] = (0.3 * df['Overall'] + 0.3 * df['Potential'] + 0.2 * df['SprintSpeed']).round(2)
print("Top 10 dupa scor general:")
print(df.nlargest(10,'Score')[['Name','Score','Overall','Potential']].to_string(index=False))
```

```python
# Ex 15 – Afaceri bune: diferenta Value - Wage (descrescator)
df_deal = df[['Name','Wage','Value']].copy()
df_deal['difference'] = df_deal['Value'] - df_deal['Wage']
df_deal_sorted = df_deal.sort_values('difference', ascending=False)
print("Top 10 afaceri bune (Value - Wage):")
print(df_deal_sorted.head(10).to_string(index=False))
```

```python
# Ex 16 – Scatter plot Value vs Wage (seaborn)
import seaborn as sns
fig, ax = plt.subplots(figsize=(8,5))
sns.scatterplot(data=df.sample(min(200,len(df)), random_state=1),
                x='Wage', y='Value', hue='Nationality', alpha=0.7, ax=ax, legend=False)
ax.set_title('Value vs Wage (sample 200 jucatori)')
ax.set_xlabel('Wage (RON)')
ax.set_ylabel('Value (RON)')
plt.tight_layout()
plt.savefig('lab4_scatter.png', dpi=80)
plt.show()
print("Grafic salvat: lab4_scatter.png")
```
