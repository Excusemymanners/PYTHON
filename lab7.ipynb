# Laborator 7 – Regresia Liniara pe setul de date Diabetes

```python
import numpy as np
import matplotlib
matplotlib.use('Agg')
import matplotlib.pyplot as plt
import pandas as pd
from sklearn.datasets import load_diabetes
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LinearRegression
from sklearn.metrics import mean_squared_error, mean_absolute_error, r2_score

# Ex 1 – Incarcam setul de date
diabetes = load_diabetes()
df = pd.DataFrame(diabetes.data, columns=diabetes.feature_names)
df['target'] = diabetes.target
print("Primele 5 randuri:")
print(df.head())
```

```python
# Ex 2 – Listam caracteristicile
print("Caracteristici disponibile:")
for f in diabetes.feature_names:
    print(f"  {f}")
```

```python
# Ex 3 – Informatii statistice
print("Statistici descriptive:")
print(df.describe().round(4))
```

```python
# Ex 4 – Histograma BMI
fig, ax = plt.subplots(figsize=(6,4))
df['bmi'].hist(bins=20, ax=ax, color='steelblue', edgecolor='black')
ax.set_title('Distributia BMI')
ax.set_xlabel('BMI (standardizat)')
ax.set_ylabel('Frecventa')
plt.tight_layout()
plt.savefig('lab7_bmi_hist.png', dpi=80)
plt.show()
```

```python
# Ex 5 – Grafic BMI si age vs target
fig, axes = plt.subplots(1, 2, figsize=(11, 4))
axes[0].scatter(df['bmi'], df['target'], alpha=0.5, color='steelblue')
axes[0].set_xlabel('BMI'); axes[0].set_ylabel('Progresie diabet')
axes[0].set_title('BMI vs Progresie')

axes[1].scatter(df['age'], df['target'], alpha=0.5, color='tomato')
axes[1].set_xlabel('Age'); axes[1].set_ylabel('Progresie diabet')
axes[1].set_title('Age vs Progresie')

plt.tight_layout()
plt.savefig('lab7_bmi_age.png', dpi=80)
plt.show()
```

```python
# Ex 6 – Regresie liniara simpla (bmi)
X_bmi = df[['bmi']].values
y     = df['target'].values

X_train, X_test, y_train, y_test = train_test_split(X_bmi, y, test_size=0.2, random_state=42)

model_bmi = LinearRegression()
model_bmi.fit(X_train, y_train)
y_pred = model_bmi.predict(X_test)

mse = mean_squared_error(y_test, y_pred)
print(f"MSE: {mse:.2f}")
print(f"Coeficient (panta): {model_bmi.coef_[0]:.4f}")
print(f"Intercept: {model_bmi.intercept_:.4f}")

fig, ax = plt.subplots(figsize=(6,4))
ax.scatter(X_test, y_test, alpha=0.5, label='Date test')
ax.plot(X_test, y_pred, color='red', linewidth=2, label='Linie regresie')
ax.set_xlabel('BMI'); ax.set_ylabel('Progresie')
ax.set_title('Regresie liniara simpla – BMI')
ax.legend()
plt.tight_layout()
plt.savefig('lab7_reg_bmi.png', dpi=80)
plt.show()
```

```python
# Ex 7 – Regresie pe bmi + bp
X_2feat = df[['bmi','bp']].values
X_train2, X_test2, y_train2, y_test2 = train_test_split(X_2feat, y, test_size=0.2, random_state=42)

model_2 = LinearRegression()
model_2.fit(X_train2, y_train2)
y_pred2 = model_2.predict(X_test2)

r2 = r2_score(y_test2, y_pred2)
mae = mean_absolute_error(y_test2, y_pred2)
mse2 = mean_squared_error(y_test2, y_pred2)

print("Coeficienti model (bmi + bp):")
for feat, coef in zip(['bmi','bp'], model_2.coef_):
    print(f"  {feat}: {coef:.4f}")
print(f"Intercept: {model_2.intercept_:.4f}")
print(f"R²:  {r2:.4f}")
print(f"MSE: {mse2:.2f}")
print(f"MAE: {mae:.2f}")
```
