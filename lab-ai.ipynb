```python
# Importam numpy pentru lucrul cu vectori si matrici.
import numpy as np
# Importam pyplot pentru grafice.
import matplotlib.pyplot as plt
# Importam modulul de dataset-uri din scikit-learn.
import sklearn.datasets as datasets
```

```python
# Importam functia care incarca dataset-ul Iris.
from sklearn.datasets import load_iris

# Incarcam dataset-ul Iris intr-un obiect de tip Bunch.
iris = load_iris()
# Extragem matricea de caracteristici (intrari).
X = iris.data
# Extragem etichetele claselor (iesiri).
y = iris.target

print("Matricea de caracteristici (X):")
print(X)
print("\nEtichetele claselor (y):")
print(y)

# Citim numarul de exemple si numarul de caracteristici din forma lui X.
n_examples, n_features = X.shape
# Afisam numarul total de exemple.
print(f"Numar de exemple: {n_examples}")
# Afisam numarul de caracteristici per exemplu.
print(f"Numar de caracteristici: {n_features}")
# Afisam numele coloanelor (atributelor).
print("Denumirile coloanelor (atributelor):", iris.feature_names)
# Afisam numele claselor tinta.
print("Numele claselor:", iris.target_names)
```

```python
# Importam functia pentru impartirea datelor in train si test.
from sklearn.model_selection import train_test_split

# Impartim datele in 80% antrenare si 20% testare.
X_train, X_test, y_train, y_test = train_test_split(
    # X reprezinta intrarile, iar y reprezinta etichetele.
    X, y, test_size=0.2, random_state=42
)

# Afisam forma subsetului X de antrenare.
print("Shape X_train:", X_train.shape)
# Afisam forma subsetului X de testare.
print("Shape X_test:", X_test.shape)
# Afisam forma subsetului y de antrenare.
print("Shape y_train:", y_train.shape)
# Afisam forma subsetului y de testare.
print("Shape y_test:", y_test.shape)
```

```python
# Importam scalerul pentru standardizarea caracteristicilor.
from sklearn.preprocessing import StandardScaler

# Cream obiectul de standardizare.
scaler = StandardScaler()
# Potrivim scalerul pe train si standardizam datele de train.
X_train_scaled = scaler.fit_transform(X_train)
# Standardizam datele de test folosind parametrizarea invatata pe train.
X_test_scaled = scaler.transform(X_test)

# Afisam primele 3 exemple inainte de scalare.
print("Primele 3 exemple din X_train inainte de scalare:")
# Afisam efectiv liniile brute din train.
print(X_train[:3])
# Afisam un separator si mesaj pentru exemplele scalate.
print("\nPrimele 3 exemple din X_train dupa scalare:")
# Afisam aceleasi 3 exemple dupa standardizare.
print(X_train_scaled[:3])
```

```python
# Importam clasificatorul K-Nearest Neighbors.
from sklearn.neighbors import KNeighborsClassifier

# Initializam modelul KNN cu k=3 vecini.
knn = KNeighborsClassifier(n_neighbors=3)
# Antrenam modelul pe datele de antrenament scalate.
knn.fit(X_train_scaled, y_train)

# Calculam acuratetea modelului pe setul de testare scalat.
test_accuracy = knn.score(X_test_scaled, y_test)
# Afisam acuratetea obtinuta.
print(f"Acuratete pe setul de testare (k=3): {test_accuracy:.4f}")
```

```python
# Construim lista valorilor k intre 1 si 15.
k_values = list(range(1, 16))
# Initializam lista in care stocam acuratetea pentru fiecare k.
accuracies = []

# Parcurgem fiecare valoare k din intervalul definit.
for k in k_values:
    # Cream un model KNN cu valoarea curenta k.
    model = KNeighborsClassifier(n_neighbors=k)
    # Antrenam modelul pe datele de train scalate.
    model.fit(X_train_scaled, y_train)
    # Calculam acuratetea pe test si o adaugam in lista.
    accuracies.append(model.score(X_test_scaled, y_test))

# Determinam cea mai mare acuratete obtinuta.
best_accuracy = max(accuracies)
# Selectam toate valorile k care ating acuratetea maxima.
best_k_candidates = [k for k, acc in zip(k_values, accuracies) if acc == best_accuracy]
# Alegem cel mai mic k dintre candidatii optimi.
best_k = min(best_k_candidates)

# Cream figura pentru graficul acuratete in functie de k.
plt.figure(figsize=(8, 5))
# Desenam curba acuratetii.
plt.plot(k_values, accuracies, marker='o')
# Setam valorile de pe axa X.
plt.xticks(k_values)
# Eticheta axei X.
plt.xlabel('Valoarea lui k')
# Eticheta axei Y.
plt.ylabel('Acuratete pe setul de testare')
# Titlul graficului.
plt.title('Impactul valorii k asupra acuratetii KNN')
# Adaugam grila pentru lizibilitate.
plt.grid(True, alpha=0.3)
# Afisam graficul.
plt.show()

# Afisam lista acuratetilor pentru toate valorile k.
print('Acurateti pentru k=1..15:', [round(acc, 4) for acc in accuracies])
# Afisam valoarea k aleasa ca optima si scorul ei.
print(f'Valoarea optima observata: k={best_k}, cu acuratete={best_accuracy:.4f}')
# Comentam criteriul de selectie folosit cand exista egalitate.
print('Comentariu: alegem cel mai mic k dintre valorile cu acuratete maxima,')
# Justificam alegerea prin simplitate si stabilitate.
print('pentru a mentine modelul simplu si a reduce riscul de netezire excesiva.')
```

```python
# Importam functiile de evaluare pentru clasificare.
from sklearn.metrics import confusion_matrix, classification_report

# Folosim best_k daca exista; altfel folosim k=5 ca fallback.
final_k = best_k if 'best_k' in globals() else 5
# Initializam modelul final cu valoarea k aleasa.
final_model = KNeighborsClassifier(n_neighbors=final_k)
# Antrenam modelul final pe setul de train scalat.
final_model.fit(X_train_scaled, y_train)
# Generam predictiile pentru setul de testare.
y_pred = final_model.predict(X_test_scaled)

# Calculam matricea de confuzie.
cm = confusion_matrix(y_test, y_pred)
# Afisam titlul sectiunii pentru matrice.
print('Matricea de confuzie:')
# Afisam matricea propriu-zisa.
print(cm)

# Afisam titlul sectiunii pentru raport.
print('\nRaport de clasificare:')
# Afisam precision, recall, f1-score si support pe fiecare clasa.
print(classification_report(y_test, y_pred, target_names=iris.target_names))

# Generam raportul in format dictionar pentru analiza ulterioara.
report_dict = classification_report(
    # Specificam etichetele reale, predictiile si numele claselor.
    y_test, y_pred, target_names=iris.target_names, output_dict=True
)
# Construim un dictionar doar cu scorurile F1 pe clasele reale.
macro_f1 = {
    # Numele clasei devine cheia, iar F1 devine valoarea.
    cls_name: metrics['f1-score']
    # Parcurgem toate intrarile din raport.
    for cls_name, metrics in report_dict.items()
    # Pastram doar intrarile care corespund claselor Iris.
    if cls_name in iris.target_names
}
# Selectam clasa cu cel mai mare scor F1.
best_class = max(macro_f1, key=macro_f1.get)

# Afisam interpretarea clasei cel mai bine prezise.
print(
    # Compunem mesajul final cu numele clasei si valoarea F1.
    f'Interpretare: clasa cel mai bine prezisa este "{best_class}" '
    f'(F1={macro_f1[best_class]:.4f}).'
)
# Verificam daca toate clasele au acelasi F1.
if len(set(macro_f1.values())) == 1:
    # Afisam mesajul de egalitate intre clase.
    print('In acest experiment, toate clasele au performanta egala dupa F1.')
```

```python
# Alegem indecsii pentru cele 2 caracteristici dorite: petal length si petal width.
feature_indices = [2, 3]
# Extragem doar cele 2 coloane din matricea completa X.
X_2f = X[:, feature_indices]

# Cream figura pentru graficul scatter.
plt.figure(figsize=(8, 5))
# Iteram prin fiecare clasa (0, 1, 2) si numele ei.
for class_idx, class_name in enumerate(iris.target_names):
    # Selectam doar exemplele care apartin clasei curente.
    mask = y == class_idx
    # Desenam punctele pentru clasa curenta.
    plt.scatter(
        # Axa X: petal length pentru clasa curenta.
        X_2f[mask, 0],
        # Axa Y: petal width pentru clasa curenta.
        X_2f[mask, 1],
        # Eticheta folosita in legenda.
        label=class_name,
        # Transparenta punctelor.
        alpha=0.8,
    )

# Eticheta axei X.
plt.xlabel('Petal length (cm)')
# Eticheta axei Y.
plt.ylabel('Petal width (cm)')
# Titlul graficului.
plt.title('Iris - Scatter pe 2 caracteristici (petala)')
# Afisam legenda claselor.
plt.legend()
# Afisam grila pentru citire mai usoara.
plt.grid(True, alpha=0.3)
# Afisam graficul.
plt.show()

# Impartim datele pe cele 2 caracteristici in train/test (80/20).
X_train_2f, X_test_2f, y_train_2f, y_test_2f = train_test_split(
    # Datele de intrare (doar 2 caracteristici), etichete, proportia de test si seed-ul fix.
    X_2f, y, test_size=0.2, random_state=42
)

# Initializam scalerul pentru standardizare.
scaler_2f = StandardScaler()
# Fit pe train + transform pe train.
X_train_2f_scaled = scaler_2f.fit_transform(X_train_2f)
# Transform pe test folosind scalerul antrenat pe train.
X_test_2f_scaled = scaler_2f.transform(X_test_2f)

# Folosim best_k daca exista deja; altfel k implicit este 3.
k_for_2f = best_k if 'best_k' in globals() else 3
# Initializam modelul KNN cu valoarea k aleasa.
knn_2f = KNeighborsClassifier(n_neighbors=k_for_2f)
# Antrenam modelul pe train-ul scalat.
knn_2f.fit(X_train_2f_scaled, y_train_2f)

# Calculam si afisam acuratetea pe setul de testare.
print(f'Acuratete KNN pe 2 caracteristici (k={k_for_2f}): {knn_2f.score(X_test_2f_scaled, y_test_2f):.4f}')

# Citim de la utilizator lungimea petalei si eliminam spatiile.
petal_length_str = input('Introdu lungimea petalei (cm): ').strip()
# Citim de la utilizator latimea petalei si eliminam spatiile.
petal_width_str = input('Introdu latimea petalei (cm): ').strip()

# Verificam daca utilizatorul a lasat vreun camp gol.
if petal_length_str == '' or petal_width_str == '':
    # Mesaj pentru input incomplet.
    print('Nu ai introdus valori. Ruleaza din nou celula si completeaza ambele campuri.')
else:
    # Incercam sa convertim valorile din text in numere reale.
    try:
        # Conversie lungime petala la float.
        petal_length = float(petal_length_str)
        # Conversie latime petala la float.
        petal_width = float(petal_width_str)

        # Construim vectorul noii flori in formatul asteptat de model.
        new_flower = np.array([[petal_length, petal_width]])
        # Standardizam noua floare cu acelasi scaler folosit la antrenare.
        new_flower_scaled = scaler_2f.transform(new_flower)
        # Prezicem indicele clasei pentru noua floare.
        predicted_class_idx = knn_2f.predict(new_flower_scaled)[0]
        # Convertim indicele in numele clasei.
        predicted_class_name = iris.target_names[predicted_class_idx]

        # Afisam clasa prezisa.
        print(f'Clasa prezisa pentru floarea introdusa: {predicted_class_name}')
    except ValueError:
        # Mesaj afisat daca inputul nu poate fi convertit la float.
        print('Valorile trebuie sa fie numerice (exemplu: 4.7 si 1.4).')
```
